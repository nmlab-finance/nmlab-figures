```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (schéma : aucune donnée externe)


import numpy as np
import pandas as pd
from matplotlib.figure import Figure
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

C, W = nm.COLORS, nm.WIDTH_PX

EA_M3_KEY = "BSI/M.U2.Y.V.M30.X.1.U2.2300.Z01.E"   # encours M3 zone euro (BCE)


def load(series_id: str, start: str | None = None, end: str | None = None) -> pd.Series:
    """Charge une série en direct : FRED, ou le portail de la BCE pour « EA_M3 »."""
    if series_id == "EA_M3":
        url = (f"https://data-api.ecb.europa.eu/service/data/{EA_M3_KEY}"
               "?format=csvdata&detail=dataonly")
        raw = pd.read_csv(url)
        s = pd.Series(raw["OBS_VALUE"].values,
                      index=pd.PeriodIndex(raw["TIME_PERIOD"], freq="M").to_timestamp())
        s = s.sort_index() / 1000.0                 # millions -> milliards d'euros
    else:
        s = nm.load_fred(series_id)
    return s.loc[start:end]


def T(d: dict, lang: str):
    """Sélectionne le jeu de libellés de la langue demandée."""
    return d[lang]


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab du chapitre (libellés selon ``lang``)."""
    fig=nm.figure(900); ax=nm.blank_axes(fig)
    d=dict(fr=("Une égalité qui ne peut pas être fausse","La vitesse n'est pas mesurée : elle est déduite pour équilibrer l'égalité.",
               "V  =  PQ / M","On ne suit pas les billets à la trace. On définit la vitesse comme le\nrapport qui rend l'égalité vraie — le résidu qui équilibre les comptes.",
               "Conséquence : MV = PQ est une identité comptable, vraie par construction,\npartout, toujours — comme « célibataire = homme non marié ». Elle n'explique rien à elle seule.",
               "Une équation qui ne peut jamais être démentie ne prédit rien : elle organise le langage, pas la causalité."),
           en=("An equality that cannot be false","Velocity is not measured: it is deduced to balance the equality.",
               "V  =  PQ / M","No one tracks banknotes. Velocity is defined as the ratio that makes\nthe equality true — the residual that balances the books.",
               "So MV = PQ is an accounting identity, true by construction, everywhere,\nalways — like \"bachelor = unmarried man.\" It explains nothing on its own.",
               "An equation that can never be refuted predicts nothing: it organizes language, not causation."))
    t=T(d,lang); nm.header(fig,t[0],t[1])
    ax.text(W*0.43,650,t[2],ha="center",va="center",fontsize=56,fontweight="bold",color=C["rose"])
    ax.text(W*0.72,655,("par définition" if lang=="fr" else "by definition"),ha="center",va="center",fontsize=23,color=C["muted"],style="italic")
    nm.card(ax,W*0.5-720,300,1440,230,edge=C["edge"],lw=2.2,radius=16,fill=C["card"])
    ax.text(W*0.5,470,t[3],ha="center",va="center",fontsize=18,color=C["text"],linespacing=1.4)
    ax.text(W*0.5,360,t[4],ha="center",va="center",fontsize=16.5,color=C["muted"],linespacing=1.4)
    nm.footer(fig,t[5]);
    return fig


fig = build_figure(LANG)
```

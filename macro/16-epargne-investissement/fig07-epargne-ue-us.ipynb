```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title="L'Europe ne manque pas d'épargne — elle manque de tuyau",
        sub='Épargne annuelle des ménages, 2022',
        bars=('Union européenne', 'États-Unis'),
        unit=' Md€',
        msg="Le carburant est là ; ce qui manque, c'est l'intermédiation qui le\ntransformerait en investissement productif plutôt qu'en dette et en pierre.",
        note="Rapport Draghi sur la compétitivité européenne (2024), d'après Eurostat et BEA.",
    ),
    "en": dict(
        title='Europe does not lack savings — it lacks the pipe',
        sub='Annual household saving, 2022',
        bars=('European Union', 'United States'),
        unit=' €bn',
        msg='The fuel is there; what is missing is the intermediation that would turn it\ninto productive investment rather than debt and bricks.',
        note='Draghi report on European competitiveness (2024), from Eurostat and BEA.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    vals = [1390, 840]
    fig = start(940); ax = nm.axes(fig, left=0.20, right=0.95, top=0.72, bottom=0.30)
    nm.header(fig, t["title"], t["sub"])
    bars = ax.barh(list(t["bars"])[::-1], vals[::-1], color=[C["amber"], C["blue"]],
                   height=0.46, zorder=3)
    for b, v in zip(bars, vals[::-1]):
        sep = " " if lang == "fr" else ","
        ax.text(v + 25, b.get_y() + b.get_height() / 2, f"{v:,}".replace(",", sep) + t["unit"],
                va="center", fontsize=27, fontweight=800, color=C["text"])
    ax.set_xlim(0, 1750); ax.set_xticks([])
    ax.tick_params(axis="y", labelsize=22, colors=C["text"], length=0, pad=16)
    for sp in ("top", "right", "bottom", "left"): ax.spines[sp].set_visible(False)
    ax2 = nm.blank_axes(fig)
    ax2.text(W / 2, 205, t["msg"], fontsize=20, color=C["muted"], ha="center",
             va="center", linespacing=1.6)
    nm.footer(fig, t["note"])
    return fig


build_figure(LANG)
```

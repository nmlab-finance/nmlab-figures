```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title="Six actifs par senior en 1990, deux aujourd'hui, 1,3 en 2050",
        sub='Personnes de 65 ans et plus pour 100 personnes de 15 à 64 ans',
        bars=('1990', '2024', '2050 (projection)'),
        ann=('≈ 6 actifs\npar senior', '≈ 2 actifs\npar senior', '≈ 1,3 actif\npar senior'),
        note='Banque mondiale (1990, 2024) ; projection OCDE (Jones, 2024). Rapporté aux 20-64 ans,\nle ratio 2024 ressort à 54,7 % : toujours vérifier la base avant de comparer deux chiffres.',
    ),
    "en": dict(
        title='Six workers per senior in 1990, two today, 1.3 by 2050',
        sub='People aged 65+ per 100 people aged 15-64',
        bars=('1990', '2024', '2050 (projected)'),
        ann=('≈ 6 workers\nper senior', '≈ 2 workers\nper senior', '≈ 1.3 workers\nper senior'),
        note='World Bank (1990, 2024); OECD projection (Jones, 2024). Against the 20-64 base the 2024\nratio is 54.7% — always check the base before comparing two figures.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    vals = [17.3, 49.2, 79.0]
    fig = start(1010); ax = nm.axes(fig, left=0.075, bottom=0.20)
    nm.header(fig, t["title"], t["sub"])
    cols = [C["blue"], C["amber"], C["rose"]]
    bars = ax.bar(t["bars"], vals, color=cols, width=0.5, zorder=3)
    for b, v, a in zip(bars, vals, t["ann"]):
        ax.text(b.get_x() + b.get_width() / 2, v + 2.5,
                f"{v:.1f}".replace(".", "," if lang == "fr" else ".") + " %",
                ha="center", va="bottom", fontsize=27, fontweight=800, color=C["text"])
        ax.text(b.get_x() + b.get_width() / 2, v / 2, a, ha="center", va="center",
                fontsize=19.5, color=C["bg"], fontweight=700, linespacing=1.5)
    ax.set_ylim(0, 92); ax.set_yticks([0, 20, 40, 60, 80])
    ax.set_yticklabels([f"{v} %" for v in (0, 20, 40, 60, 80)], fontsize=18, color=C["muted"])
    ax.tick_params(axis="x", labelsize=21, colors=C["muted"], length=0, pad=14)
    ax.grid(axis="y", color=C["grid"], lw=1.2); ax.set_axisbelow(True)
    for s in ("top", "right", "bottom", "left"): ax.spines[s].set_visible(False)
    nm.footer(fig, t["note"])
    return fig


build_figure(LANG)
```

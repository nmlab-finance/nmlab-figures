```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title="Anatomie d'une page de série FRED",
        sub='Les six champs à lire avant de faire confiance à un chiffre',
        rows=(('Observations', 'La dernière date disponible et sa valeur'), ('Units', "LE champ décisif : réel ou nominal, désaisonnalisé, annualisé —\nil n'existe pas de champ séparé pour la désaisonnalisation"), ('Frequency', 'Trimestrielle, mensuelle, quotidienne'), ('Source / Release', 'Qui produit la donnée, et dans quelle publication'), ('Updated', 'Quand elle a été révisée pour la dernière fois'), ('Next Release', 'Quand tombera le prochain chiffre')),
        note='Exemple : GDPC1 — « Billions of Chained 2017 Dollars, Seasonally Adjusted Annual Rate ».\nChaque mot de cette ligne change ce que le chiffre veut dire.',
    ),
    "en": dict(
        title='Anatomy of a FRED series page',
        sub='The six fields to read before trusting a number',
        rows=(('Observations', 'The latest date available and its value'), ('Units', 'THE decisive field: real or nominal, seasonally adjusted, annualised —\nthere is no separate field for seasonal adjustment'), ('Frequency', 'Quarterly, monthly, daily'), ('Source / Release', 'Who produces the data, and in which release'), ('Updated', 'When it was last revised'), ('Next Release', 'When the next figure lands')),
        note='Example: GDPC1 — « Billions of Chained 2017 Dollars, Seasonally Adjusted Annual Rate ».\nEvery word of that line changes what the number means.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    fig = start(1010); ax = nm.blank_axes(fig)
    nm.header(fig, t["title"], t["sub"])
    y = 700
    for i, (label, desc) in enumerate(t["rows"]):
        h = 104 if "\n" in desc else 74
        y -= h + 12
        col = C["amber"] if label == "Units" else C["edge"]
        nm.card(ax, 92, y, 1560, h, edge=col, lw=2.6 if label == "Units" else 1.6)
        ax.text(132, y + h / 2, label, fontsize=21, fontweight=800,
                color=C["amber"] if label == "Units" else C["blue"], va="center")
        ax.text(560, y + h / 2, desc, fontsize=18.5, color=C["text"] if label == "Units" else C["muted"],
                va="center", linespacing=1.5)
    nm.footer(fig, t["note"])
    return fig


build_figure(LANG)
```

```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (schéma : aucune donnée externe)


import numpy as np
import pandas as pd
from matplotlib.figure import Figure
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

C, W = nm.COLORS, nm.WIDTH_PX

EA_M3_KEY = "BSI/M.U2.Y.V.M30.X.1.U2.2300.Z01.E"   # encours M3 zone euro (BCE)


def load(series_id: str, start: str | None = None, end: str | None = None) -> pd.Series:
    """Charge une série en direct : FRED, ou le portail de la BCE pour « EA_M3 »."""
    if series_id == "EA_M3":
        url = (f"https://data-api.ecb.europa.eu/service/data/{EA_M3_KEY}"
               "?format=csvdata&detail=dataonly")
        raw = pd.read_csv(url)
        s = pd.Series(raw["OBS_VALUE"].values,
                      index=pd.PeriodIndex(raw["TIME_PERIOD"], freq="M").to_timestamp())
        s = s.sort_index() / 1000.0                 # millions -> milliards d'euros
    else:
        s = nm.load_fred(series_id)
    return s.loc[start:end]


def T(d: dict, lang: str):
    """Sélectionne le jeu de libellés de la langue demandée."""
    return d[lang]


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab du chapitre (libellés selon ``lang``)."""
    fig = nm.figure(1050); ax = nm.blank_axes(fig)
    d = dict(fr=("Où s'arrête la monnaie ?","Double test : servir de moyen de paiement, et garder une valeur nominale fixe.",
                 "DANS la monnaie",["Numéraire","Dépôts à vue","Livrets d'épargne","Dépôts à terme courts","Fonds monétaires de détail"],
                 "HORS de la monnaie",["Actions","Obligations longues","Or, matières premières","Immobilier","Cryptomonnaies"],
                 "Moyen de paiement\n+ valeur nominale fixe","Valeur qui fluctue\nou paiement indirect",
                 "L'action est liquide, mais son prix danse : c'est un placement, pas de l'argent. Réf. : FMI, BCE."),
             en=("Where does money stop?","A double test: serve as a means of payment, and keep a fixed nominal value.",
                 "INSIDE money",["Currency","Demand deposits","Savings deposits","Short time deposits","Retail money funds"],
                 "OUTSIDE money",["Stocks","Long-term bonds","Gold, commodities","Real estate","Cryptocurrencies"],
                 "Means of payment\n+ fixed nominal value","Fluctuating value\nor indirect payment",
                 "A stock is liquid, but its price dances: it is an investment, not money. Ref.: IMF, ECB."))
    t=T(d,lang); nm.header(fig,t[0],t[1])
    def panel(cx,title,items,crit,c):
        nm.card(ax,cx-380,120,760,720,edge=c,lw=2.8,radius=20,fill=C["card"])
        ax.text(cx,780,title,ha="center",va="center",fontsize=25,fontweight="bold",color=c)
        y=690
        for it in items:
            ax.text(cx-330,y,"•",ha="left",va="center",fontsize=22,color=c)
            ax.text(cx-290,y,it,ha="left",va="center",fontsize=19,color=C["text"])
            y-=88
        ax.text(cx,190,crit,ha="center",va="center",fontsize=16,color=C["muted"],style="italic",linespacing=1.25)
    panel(W*0.27,t[2],t[3],t[6],C["green"])
    panel(W*0.73,t[4],t[5],t[7],C["rose"])
    nm.footer(fig,t[8]);
    return fig


fig = build_figure(LANG)
```

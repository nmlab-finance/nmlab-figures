```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title="Trois erreurs classiques, trois remèdes d'une ligne",
        sub='Les messages rouges que tout débutant croisera — et qui ne sont jamais graves',
        h=('Le message', 'La cause', 'Le remède'),
        rows=(("ValueError: Missing column\nprovided to 'parse_dates': 'DATE'", 'FRED a renommé sa colonne de dates\nen observation_date en décembre 2024.', 'index_col=0, parse_dates=[0]'), ("could not convert string to float: '.'", 'Les valeurs manquantes sont notées par\nun point : la colonne est lue comme du texte.', 'na_values="."'), ("ModuleNotFoundError:\nNo module named 'pandas'", "La bibliothèque n'est pas installée\n(sur Colab, elle l'est déjà).", 'pip install pandas')),
        note="Le vrai débutant n'est pas celui qui se trompe : c'est celui qui abandonne à la première erreur rouge.",
    ),
    "en": dict(
        title='Three classic errors, three one-line fixes',
        sub='The red messages every beginner will meet — none of them serious',
        h=('The message', 'The cause', 'The fix'),
        rows=(("ValueError: Missing column\nprovided to 'parse_dates': 'DATE'", 'FRED renamed its date column to\nobservation_date in December 2024.', 'index_col=0, parse_dates=[0]'), ("could not convert string to float: '.'", 'Missing values are written as a dot:\nthe column is read as text.', 'na_values="."'), ("ModuleNotFoundError:\nNo module named 'pandas'", 'The library is not installed\n(on Colab it already is).', 'pip install pandas')),
        note='The real beginner is not the one who makes mistakes: it is the one who quits at the first red error.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    fig = start(1010); ax = nm.blank_axes(fig)
    nm.header(fig, t["title"], t["sub"])
    xs, ws = (92, 700, 1250), (580, 520, 402)
    for x, w_, head, col in zip(xs, ws, t["h"], (C["rose"], C["muted"], C["teal"])):
        ax.text(x + 20, 700, head, fontsize=19.5, fontweight=800, color=col, va="bottom")
    y = 672
    for msg, cause, fix in t["rows"]:
        h = 148
        y -= h + 16
        nm.card(ax, 92, y, 1560, h, edge=C["edge"])
        ax.text(112, y + h / 2, msg, fontsize=16.5, color=C["rose"], va="center",
                fontfamily="monospace", linespacing=1.6)
        ax.text(720, y + h / 2, cause, fontsize=17.5, color=C["muted"], va="center",
                linespacing=1.55)
        ax.text(1240, y + h / 2, fix, fontsize=15.5, color=C["teal"], va="center",
                fontfamily="monospace")
    nm.footer(fig, t["note"])
    return fig


build_figure(LANG)
```

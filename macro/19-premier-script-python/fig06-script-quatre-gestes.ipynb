```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title='Le script en quatre gestes',
        sub="Ce que font vraiment les huit lignes de l'atelier",
        steps=(('1', 'import pandas as pd', 'On prend les outils : pandas pour les données,\nmatplotlib pour le dessin.'), ('2', 'url = "…/fredgraph.csv?id=GDPC1"', "On colle l'identifiant FRED du chapitre 18\nà la fin de l'adresse d'export."), ('3', 'df = pd.read_csv(url, index_col=0,\n                 parse_dates=[0])', 'Une seule instruction télécharge le fichier\net le lit. Par position, pas par nom : le script\nsurvit au prochain changement de format.'), ('4', 'df.plot()', 'On dessine. Le graphique est relançable\nà volonté, modifiable, partageable.')),
        note="Chaque geste tient en une ligne — c'est ce qui rend l'analyse reproductible, au contraire du tableur.",
    ),
    "en": dict(
        title='The script in four moves',
        sub="What the workshop's eight lines actually do",
        steps=(('1', 'import pandas as pd', 'Pick up the tools: pandas for the data,\nmatplotlib for the drawing.'), ('2', 'url = "…/fredgraph.csv?id=GDPC1"', "Paste chapter 18's FRED identifier\nat the end of the export address."), ('3', 'df = pd.read_csv(url, index_col=0,\n                 parse_dates=[0])', 'One instruction downloads the file and reads it.\nBy position, not by name: the script survives\nthe next format change.'), ('4', 'df.plot()', 'Draw. The chart can be re-run at will,\nedited, shared.')),
        note='Each move fits on one line — that is what makes the analysis reproducible, unlike a spreadsheet.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    fig = start(1080); ax = nm.blank_axes(fig)
    nm.header(fig, t["title"], t["sub"])
    y = 770
    for num, code, desc in t["steps"]:
        h = 150 if "\n" in code or desc.count("\n") > 1 else 118
        y -= h + 14
        nm.card(ax, 92, y, 1560, h, edge=C["edge"])
        ax.add_patch(plt.Circle((150, y + h - 46), 26, color=C["blue"], zorder=4))
        ax.text(150, y + h - 46, num, fontsize=21, fontweight=800, color=C["bg"],
                ha="center", va="center", zorder=5)
        ax.text(200, y + h - 30, code, fontsize=19, color=C["teal"], va="top",
                fontfamily="monospace", linespacing=1.5)
        ax.text(930, y + h / 2, desc, fontsize=18, color=C["muted"], va="center",
                linespacing=1.55)
    nm.footer(fig, t["note"])
    return fig


fig = build_figure(LANG)
```

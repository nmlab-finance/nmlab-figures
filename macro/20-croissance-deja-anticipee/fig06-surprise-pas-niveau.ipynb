```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# (aucune donnée externe à charger)


from matplotlib.figure import Figure
import matplotlib.pyplot as plt
import pandas as pd

C, W = nm.COLORS, nm.WIDTH_PX


def wrap(ax, x: float, y: float, text: str, *, size: float = 19, color: str | None = None,
         weight: int = 500, ha: str = "left", va: str = "top", width: int = 42,
         lh: float = 1.5) -> int:
    """Écrit un texte replié à ``width`` caractères (coordonnées pixels)."""
    import textwrap
    lines: list[str] = []
    for para in text.split("\n"):
        lines += textwrap.wrap(para, width) or [""]
    ax.text(x, y, "\n".join(lines), fontsize=size, color=color or C["muted"],
            fontweight=weight, ha=ha, va=va, linespacing=lh, zorder=5)
    return len(lines)


def start(height: int = 1010) -> Figure:
    """Figure NMLab au format du site : 1747 px de large, fond sombre."""
    fig = nm.figure(height_px=height)
    fig.patch.set_facecolor(C["bg"])
    return fig


def dec(v: float, lang: str, n: int = 1, sign: bool = False) -> str:
    """Formate un nombre à la française (virgule, moins typographique) ou à l'anglaise."""
    s = f"{v:+.{n}f}" if sign else f"{v:.{n}f}"
    return s.replace("-", "−").replace(".", ",") if lang == "fr" else s


LABELS = {
    "fr": dict(
        title='Le marché paie la surprise, jamais le niveau',
        sub='Deux publications de croissance, deux réactions contre-intuitives',
        a_t='Un excellent chiffre… qui fait baisser',
        a_n='Publié : 2,5 %',
        a_c='Consensus : 3,0 %',
        a_r='Les prix reculent',
        b_t='Un chiffre médiocre… qui fait monter',
        b_n='Publié : 1,0 %',
        b_c='Consensus : 0,5 %',
        b_r='Les prix montent',
        msg='Ce qui était attendu était déjà payé : seule la partie inattendue reste à intégrer.',
        note="Andersen, Bollerslev, Diebold & Vega (2003) : sur les fenêtres d'annonce, seule la composante\ninattendue déplace les prix — et les mauvaises nouvelles pèsent plus lourd que les bonnes.",
    ),
    "en": dict(
        title='Markets pay for the surprise, never the level',
        sub='Two growth releases, two counter-intuitive reactions',
        a_t='An excellent figure… that pushes prices down',
        a_n='Released: 2.5%',
        a_c='Consensus: 3.0%',
        a_r='Prices fall',
        b_t='A mediocre figure… that pushes prices up',
        b_n='Released: 1.0%',
        b_c='Consensus: 0.5%',
        b_r='Prices rise',
        msg='What was expected had already been paid for: only the unexpected part is left to price in.',
        note='Andersen, Bollerslev, Diebold & Vega (2003): in announcement windows only the unexpected\ncomponent moves prices — and bad news weighs more than good.',
    ),
}


def build_figure(lang: str = "fr") -> Figure:
    """Construit la figure NMLab (libellés selon ``lang``)."""
    t = LABELS[lang]
    fig = start(1010); ax = nm.blank_axes(fig)
    nm.header(fig, t["title"], t["sub"])
    for i, (k, col, arr) in enumerate((("a", C["rose"], "↓"), ("b", C["teal"], "↑"))):
        x = 92 + i * 800
        nm.card(ax, x, 290, 760, 440, edge=col)
        wrap(ax, x + 42, 686, t[f"{k}_t"], size=22, color=C["text"], weight=800, width=34)
        ax.text(x + 42, 566, t[f"{k}_n"], fontsize=20, color=C["muted"], va="center")
        ax.text(x + 42, 512, t[f"{k}_c"], fontsize=20, color=C["muted"], va="center")
        ax.plot([x + 42, x + 718], [462, 462], color=C["edge"], lw=1.6)
        ax.text(x + 42, 380, arr, fontsize=44, color=col, va="center", fontweight=800)
        ax.text(x + 110, 380, t[f"{k}_r"], fontsize=23, color=col, va="center", fontweight=700)
    ax.text(W / 2, 218, t["msg"], fontsize=20.5, color=C["amber"], ha="center",
            va="center", fontweight=600)
    nm.footer(fig, t["note"])
    return fig


build_figure(LANG)
```

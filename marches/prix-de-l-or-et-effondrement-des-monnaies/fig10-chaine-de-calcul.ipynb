```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# Un schéma : les six étapes de l'annexe A, dans l'ordre où elles s'appliquent.
# A diagram: the six steps of appendix A, in the order they are applied.
STEPS = ["monthly", "local", "quarterly", "base", "geometric", "share"]


from matplotlib.figure import Figure
from matplotlib.patches import FancyArrowPatch

LABELS = {
    "fr": dict(
        title="La chaîne de calcul, de la série brute à la part retirée",
        sub="Les six étapes de l'annexe A — les trois familles de séries suivent le même chemin.",
        steps=dict(
            monthly=("Moyennes mensuelles", "Or au fixing de Londres,\nchange H.10 de la Réserve fédérale"),
            local=("Prix locaux", "Or en dollars × taux de change,\nzone par zone, mois par mois"),
            quarterly=("Passage au trimestre", "Moyenne des trois mois ;\naucune série n'est interpolée"),
            base=("Base 100 en 1999 T1", "Chaque série est ramenée\nà son propre point de départ"),
            geometric=("Moyenne géométrique", "Huit zones, un poids de 1/8 :\nor, prix et monnaie large"),
            share=("Parts retirées", "ln(dénominateur) ÷ ln(indice-or),\npour les prix puis la monnaie")),
        note="Les trois familles de séries — or, prix à la consommation, monnaie large — parcourent la même chaîne\n"
             "avec les mêmes poids : c'est cette symétrie qui rend les divisions finales comparables."),
    "en": dict(
        title="The computation chain, from raw series to removed share",
        sub="The six steps of appendix A — all three families of series follow the same path.",
        steps=dict(
            monthly=("Monthly averages", "Gold at the London fixing,\nFederal Reserve H.10 exchange rates"),
            local=("Local prices", "Dollar gold × exchange rate,\nzone by zone, month by month"),
            quarterly=("Move to quarters", "Average of the three months;\nno series is interpolated"),
            base=("1999 Q1 = 100", "Each series is rebased\non its own starting point"),
            geometric=("Geometric mean", "Eight zones, 1/8 weight each:\ngold, prices and broad money"),
            share=("Removed shares", "ln(denominator) ÷ ln(gold index),\nfor prices then money")),
        note="All three families of series — gold, consumer prices, broad money — run through the same chain with\n"
             "the same weights: that symmetry is what makes the final divisions comparable."),
}


def build_figure(steps: list[str], lang: str) -> Figure:
    """Six cartes numérotées en serpentin, reliées par des flèches dans l'ordre de lecture."""
    text = LABELS[lang]
    palette = [nm.COLORS["blue"], nm.COLORS["blue"], nm.COLORS["blue"],
               nm.COLORS["teal"], nm.COLORS["teal"], nm.COLORS["amber"]]
    width, height, gap = 480, 250, 71
    columns = [82, 82 + width + gap, 82 + 2 * (width + gap)]
    rows = [500, 145]

    fig = nm.figure(height_px=940)
    ax = nm.blank_axes(fig)

    boxes = []
    for rank, key in enumerate(steps):
        row, column = divmod(rank, 3)
        x = columns[column if row == 0 else 2 - column]
        y = rows[row]
        boxes.append((x, y))
        color = palette[rank]
        nm.card(ax, x, y, width, height, edge=color)
        ax.text(x + 34, y + height - 34, str(rank + 1), ha="left", va="top", fontsize=30,
                fontweight="bold", color=color)
        title, detail = text["steps"][key]
        ax.text(x + 92, y + height - 40, title, ha="left", va="top", fontsize=22.5,
                fontweight="bold", color=nm.COLORS["text"])
        ax.text(x + 34, y + 34, detail, ha="left", va="bottom", fontsize=18,
                color=nm.COLORS["muted"], linespacing=1.5)

    for rank in range(len(boxes) - 1):
        (x0, y0), (x1, y1) = boxes[rank], boxes[rank + 1]
        if y0 == y1 and x1 > x0:                      # voisines, sens de lecture
            start = (x0 + width + 12, y0 + height / 2)
            end = (x1 - 12, y1 + height / 2)
        elif y0 == y1:                                # rangée du bas, parcourue en sens inverse
            start = (x0 - 12, y0 + height / 2)
            end = (x1 + width + 12, y1 + height / 2)
        else:                                         # descente en bout de ligne
            start = (x0 + width / 2, y0 - 12)
            end = (x1 + width / 2, y1 + height + 12)
        ax.add_patch(FancyArrowPatch(start, end, arrowstyle="-|>", mutation_scale=24,
                                     color=nm.COLORS["edge"], lw=2.4, shrinkA=0, shrinkB=0))

    nm.header(fig, text["title"], text["sub"])
    nm.footer(fig, text["note"])
    return fig


fig = build_figure(STEPS, LANG)
```

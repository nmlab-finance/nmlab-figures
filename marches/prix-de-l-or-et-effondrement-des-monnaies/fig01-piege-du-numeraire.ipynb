```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# Un schéma : rien à télécharger. Chaque scénario dit combien de paniers de biens
# une once permet d'acheter aujourd'hui, et par combien les prix en dollars ont
# été multipliés entre-temps. Ce sont les deux cas PURS ; un mouvement réel se
# situe entre les deux, d'où le rail de mélange en bas de la figure.
# A diagram: nothing to download. Each scenario says how many baskets of goods one
# ounce buys today, and by how much dollar prices were multiplied. These are the two
# PURE cases; a real move sits between them — hence the blend rail at the bottom.
SCENARIOS = [("a", 2, 1.0), ("b", 1, 2.0)]


from matplotlib.colors import to_rgb
from matplotlib.figure import Figure
from matplotlib.patches import Rectangle

LABELS = {
    "fr": dict(
        title="Le même prix affiché, deux explications possibles",
        sub="Deux cas purs : dans l'un et l'autre le prix de l'once double en dollars, mais ce qu'une once permet d'acheter diffère.",
        a="Scénario A — l'or gagne en valeur",
        b="Scénario B — le dollar perd en valeur",
        start="AU DÉPART", now="AUJOURD'HUI",
        ounce="1 once\nd'or", basket="panier\nde biens",
        hint_a="Les prix des biens en dollars n'ont pas bougé.",
        hint_b="Les prix des biens en dollars ont doublé.",
        power_a="Pouvoir d'achat de l'or : ×2",
        power_b="Pouvoir d'achat de l'or : inchangé",
        price="Prix de l'or en dollars : ×2",
        mix="En pratique, ce n'est ni tout l'un ni tout l'autre",
        mix_left="tout vient de l'or", mix_mid="moitié-moitié",
        mix_right="tout vient du dollar",
        mix_note="Un mouvement réel se place quelque part sur cette ligne — et le prix affiché, seul, ne dit pas où.",
        note="Un prix est un rapport : il bouge quand l'or bouge, mais aussi quand la monnaie bouge.\n"
             "D'où la méthode de l'enquête — changer de dénominateur pour estimer la position sur cette ligne."),
    "en": dict(
        title="The same quoted price, two possible explanations",
        sub="Two pure cases: in both, the dollar price of an ounce doubles, but what one ounce buys differs.",
        a="Scenario A — gold gains value",
        b="Scenario B — the dollar loses value",
        start="AT THE START", now="TODAY",
        ounce="1 ounce\nof gold", basket="basket\nof goods",
        hint_a="The dollar prices of goods have not moved.",
        hint_b="The dollar prices of goods have doubled.",
        power_a="Purchasing power of gold: ×2",
        power_b="Purchasing power of gold: unchanged",
        price="Gold price in dollars: ×2",
        mix="In practice, it is neither one nor the other alone",
        mix_left="all from gold", mix_mid="half and half",
        mix_right="all from the dollar",
        mix_note="A real move sits somewhere along this line — and the quoted price alone does not say where.",
        note="A price is a ratio: it moves when gold moves, but also when money moves.\n"
             "Hence the method of this study — change the denominator to estimate the position along that line."),
}


def blend(start: str, middle: str, end: str, share: float) -> tuple[float, float, float]:
    """Dégradé en deux temps (``share`` de 0 à 1), chaud vers froid via une teinte neutre.

    Passer par le gris du thème évite le vert terne d'une interpolation directe
    entre l'ambre et le bleu.
    """
    left, right = (start, middle) if share < 0.5 else (middle, end)
    local = share * 2 if share < 0.5 else (share - 0.5) * 2
    return tuple(a + (b - a) * local for a, b in zip(to_rgb(left), to_rgb(right)))


def build_figure(scenarios: list[tuple[str, int, float]], lang: str) -> Figure:
    """Deux cartes symétriques — les cas purs — puis le rail continu qui les relie."""
    text = LABELS[lang]
    colors = {"a": nm.COLORS["amber"], "b": nm.COLORS["blue"]}

    fig = nm.figure(height_px=1350)
    ax = nm.blank_axes(fig)

    for (key, baskets_now, _), x0 in zip(scenarios, (82, 905)):
        accent = colors[key]
        nm.card(ax, x0, 440, 760, 740, edge=accent)
        ax.text(x0 + 44, 1140, text[key], ha="left", va="top", fontsize=22,
                fontweight="bold", color=accent)

        for row, (moment, baskets) in enumerate(((text["start"], 1), (text["now"], baskets_now))):
            y = 956 - row * 160
            ax.text(x0 + 44, y + 96, moment, ha="left", va="bottom", fontsize=16.5,
                    fontweight="bold", color=accent)
            nm.card(ax, x0 + 44, y, 200, 78, edge=accent, fill=nm.COLORS["bg"], lw=2, radius=12)
            ax.text(x0 + 144, y + 39, text["ounce"], ha="center", va="center", fontsize=19,
                    fontweight="bold", color=nm.COLORS["text"], linespacing=1.3, zorder=4)
            ax.text(x0 + 282, y + 39, "=", ha="center", va="center", fontsize=27,
                    fontweight="bold", color=nm.COLORS["muted"], zorder=4)
            for slot in range(baskets):                # une pastille par panier acheté
                bx = x0 + 320 + slot * 176
                nm.card(ax, bx, y, 160, 78, edge=nm.COLORS["edge"], fill=nm.COLORS["bg"],
                        lw=2, radius=12)
                ax.text(bx + 80, y + 39, text["basket"], ha="center", va="center", fontsize=17,
                        color=nm.COLORS["muted"], linespacing=1.3, zorder=4)

        ax.text(x0 + 44, 768, text[f"hint_{key}"], ha="left", va="top", fontsize=17.5,
                color=nm.COLORS["muted"], zorder=4)

        nm.card(ax, x0 + 44, 624, 672, 86, edge=accent, fill=nm.COLORS["bg"], lw=2, radius=14)
        ax.text(x0 + 380, 667, text[f"power_{key}"], ha="center", va="center", fontsize=20,
                fontweight="bold", color=accent, zorder=4)
        nm.card(ax, x0 + 44, 496, 672, 86, edge=accent, fill=accent, radius=14)
        ax.text(x0 + 380, 539, text["price"], ha="center", va="center", fontsize=21,
                fontweight="bold", color=nm.COLORS["bg"], zorder=4)

    # Le rail : les deux cartes ne sont que les extrémités d'un continuum.
    nm.card(ax, 82, 120, 1583, 270, edge=nm.COLORS["edge"])
    ax.text(873, 366, text["mix"], ha="center", va="top", fontsize=21,
            fontweight="bold", color=nm.COLORS["text"], zorder=4)
    left, right, steps = 126, 1621, 160
    width = (right - left) / steps
    for step in range(steps):
        ax.add_patch(Rectangle((left + step * width, 252), width * 1.02, 22, zorder=3,
                               facecolor=blend(colors["a"], nm.COLORS["muted"], colors["b"], step / (steps - 1))))
    ax.plot([873] * 2, [244, 282], color=nm.COLORS["text"], lw=2, ls=(0, (4, 3)), zorder=4)
    for x, label, ha, color in ((left, text["mix_left"], "left", colors["a"]),
                                (873, text["mix_mid"], "center", nm.COLORS["muted"]),
                                (right, text["mix_right"], "right", colors["b"])):
        ax.text(x, 232, label, ha=ha, va="top", fontsize=17.5, color=color, zorder=4)
    ax.text(873, 182, text["mix_note"], ha="center", va="top", fontsize=18.5,
            color=nm.COLORS["text"], zorder=4)

    nm.header(fig, text["title"], text["sub"])
    nm.footer(fig, text["note"])
    return fig


fig = build_figure(SCENARIOS, LANG)
```

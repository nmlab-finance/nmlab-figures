```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# Un schéma : rien à télécharger. Chaque scénario dit combien de paniers de biens
# une once permet d'acheter aujourd'hui, et par combien les prix en dollars ont
# été multipliés entre-temps.
# A diagram: nothing to download. Each scenario says how many baskets of goods one
# ounce buys today, and by how much dollar prices of goods were multiplied.
SCENARIOS = [("a", 2, 1.0), ("b", 1, 2.0)]


from matplotlib.figure import Figure

LABELS = {
    "fr": dict(
        title="Le même prix affiché, deux explications possibles",
        sub="Dans les deux cas, le prix de l'once double en dollars — mais ce qu'une once permet d'acheter n'est pas le même.",
        a="Scénario A — l'or gagne en valeur",
        b="Scénario B — chaque dollar achète moins",
        start="AU DÉPART", now="AUJOURD'HUI",
        ounce="1 once\nd'or", basket="panier\nde biens",
        hint_a="Les prix des biens en dollars n'ont pas bougé.",
        hint_b="Les prix des biens en dollars ont doublé.",
        power_a="Pouvoir d'achat de l'or : ×2",
        power_b="Pouvoir d'achat de l'or : inchangé",
        price="Prix de l'or en dollars : ×2",
        verdict="Le même prix en dollars ne dit pas lequel des deux scénarios s'est produit.",
        note="Un prix est un rapport : il bouge quand l'or bouge, mais aussi quand la monnaie bouge.\n"
             "D'où la méthode de l'enquête — changer de dénominateur plutôt que commenter le seul prix en dollars."),
    "en": dict(
        title="The same quoted price, two possible explanations",
        sub="In both cases the dollar price of an ounce doubles — but what one ounce actually buys is not the same.",
        a="Scenario A — gold gains value",
        b="Scenario B — each dollar buys less",
        start="AT THE START", now="TODAY",
        ounce="1 ounce\nof gold", basket="basket\nof goods",
        hint_a="The dollar prices of goods have not moved.",
        hint_b="The dollar prices of goods have doubled.",
        power_a="Purchasing power of gold: ×2",
        power_b="Purchasing power of gold: unchanged",
        price="Gold price in dollars: ×2",
        verdict="The same dollar price cannot tell which of the two scenarios happened.",
        note="A price is a ratio: it moves when gold moves, but also when money moves.\n"
             "Hence the method of this study — change the denominator instead of reading the dollar price alone."),
}


def build_figure(scenarios: list[tuple[str, int, float]], lang: str) -> Figure:
    """Deux cartes symétriques : même prix affiché, deux paniers de biens différents."""
    text = LABELS[lang]
    colors = {"a": nm.COLORS["amber"], "b": nm.COLORS["blue"]}

    fig = nm.figure(height_px=1180)
    ax = nm.blank_axes(fig)

    for (key, baskets_now, _), x0 in zip(scenarios, (82, 905)):
        accent = colors[key]
        nm.card(ax, x0, 250, 760, 740, edge=accent)
        ax.text(x0 + 44, 950, text[key], ha="left", va="top", fontsize=22,
                fontweight="bold", color=accent)

        for row, (moment, baskets) in enumerate(((text["start"], 1), (text["now"], baskets_now))):
            y = 766 - row * 160
            ax.text(x0 + 44, y + 96, moment, ha="left", va="bottom", fontsize=16.5,
                    fontweight="bold", color=accent)
            nm.card(ax, x0 + 44, y, 200, 78, edge=accent, fill=nm.COLORS["bg"], lw=2, radius=12)
            ax.text(x0 + 144, y + 39, text["ounce"], ha="center", va="center", fontsize=19,
                    fontweight="bold", color=nm.COLORS["text"], linespacing=1.3, zorder=4)
            ax.text(x0 + 282, y + 39, "=", ha="center", va="center", fontsize=27,
                    fontweight="bold", color=nm.COLORS["muted"], zorder=4)
            for slot in range(baskets):                # une pastille par panier acheté
                bx = x0 + 320 + slot * 176
                nm.card(ax, bx, y, 160, 78, edge=nm.COLORS["edge"], fill=nm.COLORS["bg"],
                        lw=2, radius=12)
                ax.text(bx + 80, y + 39, text["basket"], ha="center", va="center", fontsize=17,
                        color=nm.COLORS["muted"], linespacing=1.3, zorder=4)

        ax.text(x0 + 44, 578, text[f"hint_{key}"], ha="left", va="top", fontsize=17.5,
                color=nm.COLORS["muted"], zorder=4)

        nm.card(ax, x0 + 44, 434, 672, 86, edge=accent, fill=nm.COLORS["bg"], lw=2, radius=14)
        ax.text(x0 + 380, 477, text[f"power_{key}"], ha="center", va="center", fontsize=20,
                fontweight="bold", color=accent, zorder=4)
        nm.card(ax, x0 + 44, 306, 672, 86, edge=accent, fill=accent, radius=14)
        ax.text(x0 + 380, 349, text["price"], ha="center", va="center", fontsize=21,
                fontweight="bold", color=nm.COLORS["bg"], zorder=4)

    nm.card(ax, 82, 112, 1583, 100, edge=nm.COLORS["edge"])
    ax.text(873.5, 162, text["verdict"], ha="center", va="center", fontsize=22,
            color=nm.COLORS["text"], zorder=4)

    nm.header(fig, text["title"], text["sub"])
    nm.footer(fig, text["note"])
    return fig


build_figure(SCENARIOS, LANG)
```

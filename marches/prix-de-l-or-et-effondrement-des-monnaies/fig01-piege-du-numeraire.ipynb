```python
LANG = "fr"   # "fr" ou "en" — langue des libellés / label language

# Récupère puis active le style partagé NMLab (thème sombre + police Inter).
# Fetch and activate the shared NMLab style (dark theme + Inter font).
import urllib.request

urllib.request.urlretrieve("https://raw.githubusercontent.com/nmlab-finance/nmlab-figures/main/nmlab_style.py", "nmlab_style.py")
import nmlab_style as nm

nm.setup()


# Un schéma : rien à télécharger. Chaque récit est décrit par le facteur appliqué
# à la valeur de l'or et à celui du pouvoir d'achat de la monnaie.
# A diagram: nothing to download. Each story is described by the factor applied to
# the value of gold and to the purchasing power of money.
STORIES = [("a", 2.0, 1.0), ("b", 1.0, 0.5)]


from matplotlib.figure import Figure
from matplotlib.patches import FancyArrowPatch, Rectangle

LABELS = {
    "fr": dict(
        title="Le même prix affiché, deux récits incompatibles",
        sub="Une once contre 4 587 dollars en mai 2026 — ce que l'observation du prix ne permet pas de trancher.",
        fact="Ce que l'on observe", price="1 once = 4 587 dollars",
        a="Récit A — l'or est devenu plus désirable",
        b="Récit B — chaque dollar achète moins",
        gold="Valeur de l'or", money="Pouvoir d'achat du dollar",
        grown="doublée", shrunk="réduite de moitié", same="inchangée",
        start="niveau de départ",
        concl_a="L'or a doublé de valeur ; la règle n'a pas bougé.",
        concl_b="L'or n'a pas bougé ; c'est la règle qui a rétréci.",
        verdict="Dans les deux cas, le prix affiché double. Mesurer la règle avec elle-même ne les sépare pas.",
        note="C'est le piège du numéraire : un prix n'est pas une propriété d'une chose, mais un rapport entre deux\n"
             "choses. D'où la méthode du chapitre — non pas sortir du rapport, mais changer de dénominateur."),
    "en": dict(
        title="The same quoted price, two incompatible stories",
        sub="An ounce for 4,587 dollars in May 2026 — what watching the price alone cannot settle.",
        fact="What we observe", price="1 ounce = 4,587 dollars",
        a="Story A — gold has become more desirable",
        b="Story B — each dollar buys less",
        gold="Value of gold", money="Purchasing power of the dollar",
        grown="doubled", shrunk="halved", same="unchanged",
        start="starting level",
        concl_a="Gold doubled in value; the ruler did not move.",
        concl_b="Gold did not move; the ruler shrank.",
        verdict="In both cases the quoted price doubles. Measuring the ruler with itself cannot tell them apart.",
        note="This is the numeraire trap: a price is never a property of a thing, but a ratio between two things.\n"
             "Hence the method of this study — not escaping the ratio, but changing its denominator."),
}


def build_figure(stories: list[tuple[str, float, float]], lang: str) -> Figure:
    """Deux cartes symétriques : mêmes barres de départ, deux façons d'aboutir au même prix."""
    text = LABELS[lang]
    unit = 250                                    # longueur d'une barre au facteur 1
    colors = {"a": nm.COLORS["amber"], "b": nm.COLORS["blue"]}

    fig = nm.figure(height_px=1180)
    ax = nm.blank_axes(fig)

    nm.card(ax, 470, 846, 807, 152, edge=nm.COLORS["edge"])
    ax.text(873, 962, text["fact"], ha="center", va="top", fontsize=19.5, color=nm.COLORS["muted"])
    ax.text(873, 884, text["price"], ha="center", va="bottom", fontsize=33,
            fontweight="bold", color=nm.COLORS["text"])

    for (key, gold_factor, money_factor), x0 in zip(stories, (82, 905)):
        ax.add_patch(FancyArrowPatch((873 + (-190 if key == "a" else 190), 840),
                                     (x0 + 380, 800), arrowstyle="-|>", mutation_scale=24,
                                     color=nm.COLORS["edge"], lw=2.2, shrinkA=4, shrinkB=4))
        nm.card(ax, x0, 250, 760, 540, edge=colors[key])
        ax.text(x0 + 44, 736, text[key], ha="left", va="top", fontsize=23.5,
                fontweight="bold", color=colors[key])

        for row, (which, factor) in enumerate((("gold", gold_factor), ("money", money_factor))):
            y = 566 - row * 148
            ax.text(x0 + 44, y + 62, text[which], ha="left", va="bottom", fontsize=19.5,
                    color=nm.COLORS["muted"])
            ax.add_patch(Rectangle((x0 + 44, y), unit * factor, 40, zorder=3,
                                   facecolor=colors[key] if factor != 1 else nm.COLORS["edge"]))
            ax.plot([x0 + 44 + unit] * 2, [y - 16, y + 56], color=nm.COLORS["text"],
                    lw=2, ls=(0, (4, 3)), zorder=4)
            change = text["same"] if factor == 1 else (text["grown"] if factor > 1 else text["shrunk"])
            ax.text(x0 + 44 + unit * max(factor, 1.0) + 24, y + 20, change, ha="left", va="center",
                    fontsize=19.5, fontweight="bold" if factor != 1 else "normal",
                    color=nm.COLORS["text"] if factor != 1 else nm.COLORS["muted"])

        ax.text(x0 + 44 + unit, 396, text["start"], ha="center", va="top", fontsize=16.5,
                color=nm.COLORS["muted"], alpha=0.9)
        ax.text(x0 + 44, 296, text[f"concl_{key}"], ha="left", va="bottom", fontsize=20,
                color=nm.COLORS["text"])

    nm.card(ax, 82, 110, 1583, 96, edge=nm.COLORS["edge"], fill=nm.COLORS["card"])
    ax.text(873, 158, text["verdict"], ha="center", va="center", fontsize=22,
            color=nm.COLORS["text"])

    nm.header(fig, text["title"], text["sub"])
    nm.footer(fig, text["note"])
    return fig


build_figure(STORIES, LANG)
```
